```python
import pandas as pd

# Assuming your DataFrame is named 'df' and has a column 'arrival_date_month' with month names

# Create a dictionary mapping month names to month numbers
month_mapping = {
    'January': 1,
    'February': 2,
    'March': 3,
    'April': 4,
    'May': 5,
    'June': 6,
    'July': 7,
    'August': 8,
    'September': 9,
    'October': 10,
    'November': 11,
    'December': 12
}

# Use the map function to apply the mapping to the 'arrival_date_month' column
df['arrival_date_month_num'] = df['arrival_date_month'].map(month_mapping)
```

```python
df.head()
```

```python
df.drop(['arrival_date_month'], axis=1, inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
grouped_df = df.groupby(['arrival_date_year', 'arrival_date_month_num', 'arrival_date_day_of_month'])
aggregated_df = grouped_df[['children','babies','adults']].sum().reset_index()
```

```python
aggregated_df.head()
```

```python
aggregated_df.tail()
```

```python
aggregated_df['date'] = pd.to_datetime(aggregated_df[['arrival_date_year', 'arrival_date_month_num', 'arrival_date_day_of_month']].rename(columns={
    'arrival_date_year': 'year',
    'arrival_date_month_num': 'month',
    'arrival_date_day_of_month': 'day'
}), errors='coerce')
aggregated_df = aggregated_df.dropna(subset=['date'])
```

```python
aggregated_df.head()
```

```python
train_data = aggregated_df[(aggregated_df['date'] >= '2015-07-01') & (aggregated_df['date'] <= '2016-07-01')]
test_data = aggregated_df[(aggregated_df['date'] >= '2016-07-02') & (aggregated_df['date'] <= '2016-07-19')]
```

```python
x_train = train_data[['arrival_date_year','arrival_date_month_num','arrival_date_day_of_month']]
y_train = train_data[['children','babies','adults']]
x_test = test_data[['arrival_date_year','arrival_date_month_num','arrival_date_day_of_month']]
y_test = test_data[['children','babies','adults']]
```

```python
import xgboost as xgb
from sklearn.metrics import mean_squared_error
reg = xgb.XGBRegressor(n_estimators=1000,arly_stopping_rounds=50)
reg.fit(x_train, y_train,
        eval_set=[(x_train, y_train), (x_test, y_test)],
       verbose=False)
```

```python
y_pred = reg.predict([[2024,5,11]])
```

```python
for i in range(len(y_pred[0])):
    print(int(y_pred[0][i]))
```
